## Endpoints

1) https://pokeapi.co/api/v2/pokemon/charizard 
-  The route parameter is just the pokemons name and it gives information about that pokemon

2) https://pokeapi.co/api/v2/type/fire
- The route parameter is just the types name (Ex: Fire). It shares information about pokemon that are that same type, it gives information on the damage this type has on other types. 

```python
from pathlib import Path
import json
import pandas as pd

FIXTURE_DIR = Path("api-fixtures")
```

```python
charizard_path = FIXTURE_DIR / "charizard.json"
with charizard_path.open(encoding="utf-8") as handle:
    charizard = json.load(handle)
print(charizard.keys())
```

```python
fire_path = FIXTURE_DIR / "fire.json"
with fire_path.open(encoding="utf-8") as handle:
    fire = json.load(handle)
print(fire.keys())
```
